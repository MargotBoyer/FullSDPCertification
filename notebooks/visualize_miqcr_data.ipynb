```python
import os, sys, json, glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

sys.path.insert(0, os.path.join(os.path.dirname(os.getcwd()), 'src'))
from fastsdp_tools import get_project_path
```

## 1. Sélection du run et de l'échantillon

```python
# ── À MODIFIER ──────────────────────────────────────────────────────────────
SAMPLE_INDEX = 0   # numéro de l'échantillon (data_index)
RUN_DIR = "results/benchmark/blob_nn_4x10-1/2026_07_22_10h59_00s_MIQCR_conic_bundle"
# ────────────────────────────────────────────────────────────────────────────


sample_dir = os.path.join(get_project_path(RUN_DIR), f'sample_{SAMPLE_INDEX}')

assert os.path.isdir(sample_dir), f'Dossier introuvable : {sample_dir}'
print('Sample :', sample_dir)
```

## 2. Chargement des données

```python
with open(os.path.join(sample_dir, 'miqcr_scalars.json')) as f:
    scalars = json.load(f)

with open(os.path.join(sample_dir, 'flat_order.json')) as f:
    flat_order = [tuple(x) for x in json.load(f)]

npz = np.load(os.path.join(sample_dir, 'miqcr_matrices.npz'))
A, b = npz['A'], npz['b']
D, e = npz['D'], npz['e']
Aq, bq = npz['Aq'], npz['bq']
Dq, eq = npz['Dq'], npz['eq']
Q, c   = npz['Q'], npz['c']
u, l   = npz['u'], npz['l']

print('Scalaires du problème :')
for k, v in scalars.items():
    print(f'  {k} = {v}')
```

## 3. Résumé des matrices

```python
def matrix_summary(name, arr):
    if arr.size == 0:
        return {'name': name, 'shape': str(arr.shape), 'nnz': 0,
                'min': None, 'max': None, 'mean': None, 'std': None}
    nz = np.count_nonzero(arr)
    return {
        'name':  name,
        'shape': str(arr.shape),
        'nnz':   nz,
        'sparsity': f'{100*(1 - nz/arr.size):.1f}%',
        'min':   float(arr.min()),
        'max':   float(arr.max()),
        'mean':  float(arr.mean()),
        'std':   float(arr.std()),
    }

rows = [
    matrix_summary('Q (hessienne obj)', Q),
    matrix_summary('c (linéaire obj)',  c),
    matrix_summary('u (bornes sup)',    u),
    matrix_summary('l (bornes inf)',    l),
    matrix_summary('A (éga. lin)',      A),
    matrix_summary('b (rhs A)',         b),
    matrix_summary('D (inéga. lin)',    D),
    matrix_summary('e (rhs D)',         e),
    matrix_summary('Aq (éga. quad)',    Aq),
    matrix_summary('bq (rhs Aq)',       bq),
    matrix_summary('Dq (inéga. quad)', Dq),
    matrix_summary('eq (rhs Dq)',       eq),
]
pd.DataFrame(rows).set_index('name')
```

## 4. Heatmaps des matrices 2D

```python
def heatmap(ax, mat, title, symmetric_cmap=True):
    if mat.size == 0:
        ax.text(0.5, 0.5, '(vide)', ha='center', va='center', transform=ax.transAxes)
        ax.set_title(title)
        return
    if symmetric_cmap:
        vmax = np.abs(mat).max() or 1
        im = ax.imshow(mat, cmap='RdBu_r', vmin=-vmax, vmax=vmax, aspect='auto')
    else:
        im = ax.imshow(mat, cmap='viridis', aspect='auto')
    plt.colorbar(im, ax=ax, fraction=0.03)
    ax.set_title(title)
    ax.set_xlabel('variables')
    ax.set_ylabel('contraintes')

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
heatmap(axes[0, 0], Q,             f'Q  ({Q.shape[0]}×{Q.shape[1]})')
heatmap(axes[0, 1], A if A.size else A, f'A  ({A.shape[0]}×{A.shape[1] if A.ndim>1 else 0})')
heatmap(axes[1, 0], D if D.size else D, f'D  ({D.shape[0]}×{D.shape[1] if D.ndim>1 else 0})')

# Bornes empilées
bounds = np.stack([l, u], axis=0)
heatmap(axes[1, 1], bounds, 'Bornes l (ligne 0) et u (ligne 1)', symmetric_cmap=False)
axes[1, 1].set_yticks([0, 1])
axes[1, 1].set_yticklabels(['l', 'u'])

plt.suptitle(f'Matrices linéaires — sample {SAMPLE_INDEX}', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Heatmaps des contraintes quadratiques (une par contrainte)
def plot_quad_matrices(mats, rhs, label):
    if mats.size == 0:
        print(f'{label} : aucune contrainte quadratique.')
        return
    n_mats = min(len(mats), 6)   # affiche au plus 6
    ncols = min(n_mats, 3)
    nrows = (n_mats + ncols - 1) // ncols
    fig, axes = plt.subplots(nrows, ncols, figsize=(6*ncols, 5*nrows), squeeze=False)
    for k in range(n_mats):
        ax = axes[k // ncols][k % ncols]
        vmax = np.abs(mats[k]).max() or 1
        im = ax.imshow(mats[k], cmap='RdBu_r', vmin=-vmax, vmax=vmax, aspect='auto')
        plt.colorbar(im, ax=ax, fraction=0.03)
        ax.set_title(f'{label}[{k}]  rhs={rhs[k]:.4g}')
    for k in range(n_mats, nrows*ncols):
        axes[k // ncols][k % ncols].axis('off')
    if len(mats) > 6:
        plt.suptitle(f'{label} — 6 premières sur {len(mats)}', fontsize=13)
    plt.tight_layout()
    plt.show()

plot_quad_matrices(Aq, bq, 'Aq (égalités quad)')
plot_quad_matrices(Dq, eq, 'Dq (inégalités quad)')
```

## 5. Patterns de sparsité

```python
def sparsity_pattern(ax, mat, title):
    if mat.size == 0:
        ax.text(0.5, 0.5, '(vide)', ha='center', va='center', transform=ax.transAxes)
        ax.set_title(title)
        return
    ax.spy(mat, markersize=1, color='steelblue')
    nnz = np.count_nonzero(mat)
    ax.set_title(f'{title}\nnnz={nnz} / {mat.size}  ({100*nnz/mat.size:.1f}% dense)')

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sparsity_pattern(axes[0], Q, f'Q  {Q.shape}')
sparsity_pattern(axes[1], A, f'A  {A.shape}')
sparsity_pattern(axes[2], D, f'D  {D.shape}')
plt.suptitle(f'Patterns de sparsité — sample {SAMPLE_INDEX}', fontsize=13)
plt.tight_layout()
plt.show()
```

## 6. Distributions des valeurs non nulles

```python
def hist_nonzero(ax, arr, title, bins=50, color='steelblue'):
    vals = arr.ravel()
    nz = vals[vals != 0]
    if len(nz) == 0:
        ax.text(0.5, 0.5, '(tout nul)', ha='center', va='center', transform=ax.transAxes)
        ax.set_title(title)
        return
    ax.hist(nz, bins=bins, color=color, edgecolor='white', linewidth=0.3)
    ax.axvline(0, color='red', linewidth=0.8, linestyle='--')
    ax.set_title(f'{title}\n(n={len(nz)} valeurs non nulles)')
    ax.set_xlabel('valeur')
    ax.set_ylabel('fréquence')

arrays = [
    (Q.ravel(), 'Q', 'steelblue'),
    (c,         'c', 'darkorange'),
    (u,         'u (bornes sup)', 'seagreen'),
    (l,         'l (bornes inf)', 'tomato'),
    (A.ravel(), 'A', 'mediumpurple'),
    (b,         'b (rhs A)', 'orchid'),
    (D.ravel(), 'D', 'goldenrod'),
    (e,         'e (rhs D)', 'peru'),
]

ncols = 4
nrows = (len(arrays) + ncols - 1) // ncols
fig, axes = plt.subplots(nrows, ncols, figsize=(5*ncols, 4*nrows))
axes = axes.ravel()
for ax, (arr, title, color) in zip(axes, arrays):
    hist_nonzero(ax, arr, title, color=color)
for ax in axes[len(arrays):]:
    ax.axis('off')

plt.suptitle(f'Distributions des valeurs — sample {SAMPLE_INDEX}', fontsize=14)
plt.tight_layout()
plt.show()
```

## 7. Flat order — répartition par couche

```python
layers = [layer for (layer, j) in flat_order]
neurons = [j for (layer, j) in flat_order]

layer_counts = pd.Series(layers).value_counts().sort_index()
print('Variables par couche :')
print(layer_counts.to_string())
print(f'Total : {len(flat_order)} variables')

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].bar(layer_counts.index, layer_counts.values, color='steelblue', edgecolor='white')
axes[0].set_xlabel('Couche')
axes[0].set_ylabel('Nombre de variables')
axes[0].set_title('Variables actives par couche')
axes[0].set_xticks(layer_counts.index)

axes[1].scatter(range(len(flat_order)), layers, s=2, alpha=0.5, color='steelblue')
axes[1].set_xlabel('Indice plat (flat index)')
axes[1].set_ylabel('Couche')
axes[1].set_title('Appartenance de chaque variable à sa couche')

plt.suptitle(f'Flat order — sample {SAMPLE_INDEX}', fontsize=13)
plt.tight_layout()
plt.show()
```

## 8. Comparaison multi-échantillons (résumé scalaires)

```python
# Charge les scalaires de tous les échantillons du run sélectionné
all_samples = sorted(glob.glob(os.path.join(run_dir, 'sample_*')))
rows = []
for s in all_samples:
    scalar_file = os.path.join(s, 'miqcr_scalars.json')
    if not os.path.exists(scalar_file):
        continue
    with open(scalar_file) as f:
        sc = json.load(f)
    sc['sample'] = int(os.path.basename(s).replace('sample_', ''))
    rows.append(sc)

if rows:
    df_all = pd.DataFrame(rows).set_index('sample').sort_index()
    display(df_all)

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    for ax, col, color in zip(axes, ['n', 'm', 'p'], ['steelblue', 'seagreen', 'tomato']):
        if col in df_all.columns:
            axes_i = ax
            axes_i.bar(df_all.index, df_all[col], color=color, edgecolor='white')
            axes_i.set_title(col)
            axes_i.set_xlabel('sample index')
    plt.suptitle('Dimensions du problème par échantillon', fontsize=13)
    plt.tight_layout()
    plt.show()
else:
    print('Aucun échantillon trouvé dans ce run.')
```

## 9. Résultats du run (results.csv)

```python
csv_path = os.path.join(run_dir, 'results.csv')
if os.path.exists(csv_path):
    df_res = pd.read_csv(csv_path)
    display(df_res)

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    if 'sol_sdp' in df_res.columns:
        axes[0].bar(df_res['data_index'], df_res['sol_sdp'], color='steelblue')
        axes[0].axhline(0, color='red', linestyle='--', linewidth=0.8)
        axes[0].set_title('Valeur objective SDP par échantillon')
        axes[0].set_xlabel('data_index')
        axes[0].set_ylabel('sol_sdp')
    if 'time_build_miqcr' in df_res.columns and 'time_conic_bundle' in df_res.columns:
        width = 0.35
        x = df_res['data_index']
        axes[1].bar(x - width/2, df_res['time_build_miqcr'], width, label='extraction', color='seagreen')
        axes[1].bar(x + width/2, df_res['time_conic_bundle'],      width, label='conic bundle', color='tomato')
        axes[1].set_title('Temps par échantillon')
        axes[1].set_xlabel('data_index')
        axes[1].set_ylabel('secondes')
        axes[1].legend()
    plt.tight_layout()
    plt.show()
else:
    print(f'Pas encore de results.csv dans {run_dir}')
```
